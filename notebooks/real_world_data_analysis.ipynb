# Real-World Data Project: Retail Sales Analytics & Revenue Prediction
**Objectives:**
- Analyze a domain-specific Retail & E-Commerce dataset (`data/retail_sales_data.csv`)
- Perform end-to-end exploratory data analysis (EDA) and statistical aggregation
- Train and evaluate supervised regression models to predict order revenue
- Present actionable business findings with visualizations and conclusions

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/retail_sales_data.csv', parse_dates=['Order_Date'])
df.head()
```

```python
# Summary Statistics by Product Category
summary = df.groupby('Category')[['Revenue', 'Profit', 'Quantity', 'Customer_Rating']].agg({'Revenue': ['count', 'sum', 'mean'], 'Profit': 'sum', 'Customer_Rating': 'mean'})
summary
```

```python
# Visualize Category Revenue and Discount Impact
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df.groupby('Category')['Revenue'].sum().sort_values().plot(kind='barh', ax=axes[0], color='steelblue', title='Total Revenue by Category ($)')
df['Profit_Margin_Pct'] = (df['Profit'] / df['Revenue']) * 100
sns.lineplot(data=df, x='Discount_Rate', y='Profit_Margin_Pct', ax=axes[1], color='crimson')
axes[1].set_title('Discount Rate vs. Profit Margin (%)')
plt.tight_layout()
plt.show()
```
